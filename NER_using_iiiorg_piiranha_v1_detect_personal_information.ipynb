```python
from transformers import AutoTokenizer, AutoModelForTokenClassification
from transformers import pipeline
```

```python
tokenizer = AutoTokenizer.from_pretrained("iiiorg/piiranha-v1-detect-personal-information")
model = AutoModelForTokenClassification.from_pretrained("iiiorg/piiranha-v1-detect-personal-information")
```

```python
nlp = pipeline("ner", model=model, tokenizer=tokenizer)
```

```python
def NER(text):
  return nlp(text)
```

```python
NER("My name is Alice. You can email me at alice@example.com or call me at 123-456-7890.")
```

```python
NER("John Doe's credit card number is 1234-5678-9012-3456. His birth date is 10/10/1980.")
```

```python
NER("You can reach me at james_smith99@yahoo.com or visit my website at https://www.jamessmith.com.")
```
